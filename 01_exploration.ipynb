```python
print("Dimensions :", df.shape)

print("\nColonnes :")
print(df.columns.tolist())

print("\nValeurs manquantes :")
print(df.isna().sum())
```

```python
print("Liste complète des colonnes :")
for colonne in df.columns:
    print(colonne)

print("\nPremières lignes du tableau :")
display(df.head())

colonnes_utiles = [
    "Date",
    "Heure",
    "Date et Heure",
    "Consommation (MW)",
    "Prévision J-1 (MW)",
    "Prévision J (MW)",
    "Eolien (MW)",
    "Solaire (MW)"
]

display(df[colonnes_utiles].head(10))
```

```python
# Vérifier les valeurs manquantes
print("Valeurs manquantes :")

display(df[colonnes_utiles].isna().sum())

# Vérifier la période couverte par les données
print("Première date :", df["Date"].min())
print("Dernière date :", df["Date"].max())

# Vérifier le nombre de dates différentes
print("Nombre de dates :", df["Date"].nunique())
```

```python
# Examiner les catégories des données
print("Périmètres disponibles :")
print(df["Périmètre"].value_counts())

print("\nNatures disponibles :")
print(df["Nature"].value_counts())

# Comparer les valeurs manquantes selon la nature
display(
    df.groupby("Nature")[
        ["Consommation (MW)", "Eolien (MW)", "Solaire (MW)"]
    ].agg(["count", "size"])
)
```

```python
# Colonnes à examiner
colonnes_energie = [
    "Consommation (MW)",
    "Eolien (MW)",
    "Solaire (MW)"
]

# Compter les valeurs présentes sur chaque ligne
df["nb_valeurs_energie"] = df[colonnes_energie].notna().sum(axis=1)

# Résultat par catégorie de données
display(
    df.groupby(["Nature", "nb_valeurs_energie"]).size().unstack(fill_value=0)
)
```

```python
# Lignes où les trois valeurs sont présentes
print("Exemples avec les 3 valeurs présentes :")

display(
    df[df["nb_valeurs_energie"] == 3][
        ["Date", "Heure", "Nature"] + colonnes_energie
    ].head(5)
)

# Lignes où les trois valeurs sont absentes
print("Exemples avec les 3 valeurs manquantes :")

display(
    df[df["nb_valeurs_energie"] == 0][
        ["Date", "Heure", "Nature"] + colonnes_energie
    ].head(5)
)
```

```python
# 1. Supprimer les lignes vides (créneaux :15 et :45)
df = df.dropna(subset=["Consommation (MW)"]).copy()

# 2. Créer une vraie colonne date-heure
df["datetime"] = pd.to_datetime(df["Date"] + " " + df["Heure"])

# 3. Vérifier la période de chaque catégorie
print(df.groupby("Nature")["datetime"].agg(["min", "max", "count"]))

# 4. Vérifier qu'il n'y a pas de doublons
print("Doublons :", df.duplicated(subset=["datetime"]).sum())
```

```python
import matplotlib.pyplot as plt

# On travaille sur les données définitives (2012-2025)
dfd = df[df["Nature"] == "Données définitives"].copy()

# Colonnes utiles
dfd["heure_jour"] = dfd["datetime"].dt.hour
dfd["mois"] = dfd["datetime"].dt.month

# Graphique 1 : consommation moyenne selon l'heure de la journée
dfd.groupby("heure_jour")["Consommation (MW)"].mean().plot(
    marker="o", title="Consommation moyenne par heure"
)
plt.xlabel("Heure")
plt.ylabel("MW")
plt.show()

# Graphique 2 : consommation moyenne selon le mois
dfd.groupby("mois")["Consommation (MW)"].mean().plot(
    kind="bar", title="Consommation moyenne par mois"
)
plt.xlabel("Mois")
plt.ylabel("MW")
plt.show() 
```

```python
# Jour de la semaine : 0 = lundi, 6 = dimanche
dfd["jour_semaine"] = dfd["datetime"].dt.dayofweek
dfd["weekend"] = dfd["jour_semaine"] >= 5

# Graphique 3 : consommation moyenne par jour de la semaine
dfd.groupby("jour_semaine")["Consommation (MW)"].mean().plot(
    kind="bar", title="Consommation moyenne par jour de la semaine"
)
plt.show()

# Graphique 4 : courbe journalière, semaine vs week-end
(dfd.groupby(["heure_jour", "weekend"])["Consommation (MW)"]
    .mean().unstack().plot(title="Semaine vs week-end"))
plt.legend(["Semaine", "Week-end"])
plt.show()
```

```python
s = dfd.sort_values("datetime").set_index("datetime")["Consommation (MW)"]

base_veille = s.shift(48)       # même heure, la veille (48 pas de 30 min)
base_semaine = s.shift(48 * 7)  # même heure, la semaine précédente

test = s["2025"]
for nom, base in [("veille", base_veille), ("semaine précédente", base_semaine)]:
    erreur = (test - base["2025"]).abs()
    print(nom, "| erreur moyenne :", round(erreur.mean()), "MW |",
          "erreur relative :", round((erreur / test).mean() * 100, 1), "%")
```
